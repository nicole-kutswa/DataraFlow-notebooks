- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.n= n
    def __iter__(self):
        return self
    def __next__(self):
        
        if self.n>=1:
            current= self.n
            self.n-=1
            return current
        else:
            raise StopIteration

x= CountDown(5)
for number in x:
    print(number)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename
        self.file = open(filename, "r")

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()

        if line == "":
            self.file.close()
            raise StopIteration

        return line.strip()


# Test the iterator
reader = FileLineReader("sample.txt")

for line in reader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math


# Abstract class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


# Rectangle subclass
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


# Circle subclass
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


# Create a list of different shapes
shapes = [
    Rectangle(10, 5),
    Circle(7),
    Rectangle(4, 3),
    Circle(3)
]


# Loop through the shapes
for shape in shapes:
    print("Area:", shape.area())
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.previous = 0
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.previous > self.n:
            raise StopIteration

        number = self.previous

        self.previous, self.current = self.current, self.previous + self.current

        return number


x = Fibonacci(6)

for i in x:
    print(i)
````

# Library Management System  
  
## Scenario  
You are building a small library system to manage books, members, and borrowing operations. The system should support multiple types of books, track borrowed books, and allow iteration over members and books. The system should also demonstrate proper use of scopes and encapsulation.  
  
## Requirements and Concepts to Cover  
  
### 1. OOP & Inheritance  
- Create a base class `Book` with attributes: `title`, `author`, `year`.  
- Create subclasses:  
  - `EBook` with `file_size` and `format`.  
  - `PrintedBook` with `weight` and `dimensions`.  
- Create a base class `Member` with `name` and `member_id`.  
- Create subclasses:  
  - `StudentMember` with `grade_level`.  
  - `TeacherMember` with `department`.  
  
### 2. Polymorphism  
- Define a method `get_description()` in `Book` and override it in both `EBook` and `PrintedBook`.  
- Define a method `borrow_book()` in `Member` that behaves differently for `StudentMember` and `TeacherMember` (e.g., different borrowing limits).  
  
### 3. Iterators  
- Implement an iterator for the `Library` class to iterate over all books.  
- Implement an iterator for the `Member` class to iterate over borrowed books.  
  
### 4. Scope  
- Use private and protected attributes to demonstrate scope:  
  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
# ==========================================
# LIBRARY MANAGEMENT SYSTEM
# ==========================================

# ------------------------------------------
# 1. BOOK CLASSES
# ------------------------------------------

class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_available = True

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"


class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return (
            f"EBook: {self.title} by {self.author} "
            f"({self.year}) - {self.file_size} MB, {self.format} format"
        )


class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return (
            f"Printed Book: {self.title} by {self.author} "
            f"({self.year}) - {self.weight} kg, "
            f"dimensions: {self.dimensions}"
        )


# ------------------------------------------
# 2. MEMBER CLASSES
# ------------------------------------------

class Member:
    # Class-level variable
    total_borrowed_books = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id

        # Protected attribute
        self._borrowed_books = []

    def borrow_book(self, book):
        print("A member is borrowing a book.")

    def return_book(self, book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            book.is_available = True
            print(f"{book.title} has been returned.")
        else:
            print("This book was not borrowed by this member.")

    def get_borrowed_books(self):
        return self._borrowed_books

    # Iterator
    def __iter__(self):
        return iter(self._borrowed_books)


class StudentMember(Member):
    BORROWING_LIMIT = 3

    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level

    def borrow_book(self, book):
        if not book.is_available:
            print(f"Sorry, '{book.title}' is already borrowed.")
            return

        if len(self._borrowed_books) >= self.BORROWING_LIMIT:
            print(
                f"{self.name} has reached the student borrowing limit "
                f"of {self.BORROWING_LIMIT} books."
            )
            return

        self._borrowed_books.append(book)
        book.is_available = False

        Member.total_borrowed_books += 1

        print(f"{self.name} borrowed '{book.title}'.")


class TeacherMember(Member):
    BORROWING_LIMIT = 5

    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department

    def borrow_book(self, book):
        if not book.is_available:
            print(f"Sorry, '{book.title}' is already borrowed.")
            return

        if len(self._borrowed_books) >= self.BORROWING_LIMIT:
            print(
                f"{self.name} has reached the teacher borrowing limit "
                f"of {self.BORROWING_LIMIT} books."
            )
            return

        self._borrowed_books.append(book)
        book.is_available = False

        Member.total_borrowed_books += 1

        print(f"{self.name} borrowed '{book.title}'.")


# ------------------------------------------
# 3. LIBRARY CLASS
# ------------------------------------------

class Library:

    def __init__(self, library_name):
        # Private attribute
        self.__library_name = library_name

        self.books = []
        self.members = []

    # Getter for private attribute
    def get_library_name(self):
        return self.__library_name

    # Setter for private attribute
    def set_library_name(self, new_name):
        self.__library_name = new_name

    # Add a book
    def add_book(self, book):
        self.books.append(book)

    # Add a member
    def add_member(self, member):
        self.members.append(member)

    # Iterator for books
    def __iter__(self):
        return iter(self.books)

    # Search by title or author
    def search_books(self, search_term):
        results = []

        for book in self.books:
            if (
                search_term.lower() in book.title.lower()
                or search_term.lower() in book.author.lower()
            ):
                results.append(book)

        return results


# ==========================================
# 4. TEST THE SYSTEM
# ==========================================

# Create a library
library = Library("Kenyatta University Library")


# Create books
book1 = EBook(
    "Python Basics",
    "Nicole Kutswa",
    2024,
    5.2,
    "PDF"
)

book2 = PrintedBook(
    "Data Science Handbook",
    "Newton Munene",
    2023,
    1.2,
    "24 x 16 x 3 cm"
)

book3 = EBook(
    "Artificial Intelligence",
    "William Namara",
    2025,
    7.5,
    "EPUB"
)

book4 = PrintedBook(
    "Database Systems",
    "Bianca Omitsi",
    2022,
    1.5,
    "25 x 18 x 4 cm"
)


# Add books to library
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
library.add_book(book4)


# Create members
student = StudentMember(
    "Nicole",
    "ST001",
    "Year 4"
)

teacher = TeacherMember(
    "Dr. Kamau",
    "TE001",
    "Information Technology"
)


# Add members to library
library.add_member(student)
library.add_member(teacher)


# ==========================================
# 5. DISPLAY BOOK DESCRIPTIONS
# ==========================================

print("\n--- BOOKS IN LIBRARY ---")

for book in library:
    print(book.get_description())


# ==========================================
# 6. BORROW BOOKS
# ==========================================

print("\n--- BORROWING BOOKS ---")

student.borrow_book(book1)
student.borrow_book(book2)

teacher.borrow_book(book3)


# ==========================================
# 7. ITERATE THROUGH BORROWED BOOKS
# ==========================================

print("\n--- STUDENT'S BORROWED BOOKS ---")

for book in student:
    print(book.title)


print("\n--- TEACHER'S BORROWED BOOKS ---")

for book in teacher:
    print(book.title)


# ==========================================
# 8. SEARCH BOOKS
# ==========================================

print("\n--- SEARCH RESULTS ---")

results = library.search_books("Python")

for book in results:
    print(book.get_description())


# ==========================================
# 9. CHECK BOOK AVAILABILITY
# ==========================================

print("\n--- BOOK AVAILABILITY ---")

for book in library:
    status = "Available" if book.is_available else "Borrowed"
    print(f"{book.title}: {status}")


# ==========================================
# 10. RETURN A BOOK
# ==========================================

print("\n--- RETURNING BOOK ---")

student.return_book(book1)


# ==========================================
# 11. TOTAL BOOKS BORROWED
# ==========================================

print("\n--- TOTAL BORROWED ---")

print(
    f"Total books borrowed: "
    f"{Member.total_borrowed_books}"
)
````
