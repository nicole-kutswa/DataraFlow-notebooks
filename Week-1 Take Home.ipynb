1. **Student Grades System**  
   Write a program to accept marks for 5 subjects, calculate the average, and display the grade based on:  
   - A: 80–100  
   - B: 60–79  
   - C: 40–59  
   - F: < 40  

2. **Simple Calculator**

3. **Shopping Cart**  
   Create a program where a user can add items to a shopping cart (dictionary with item names and prices), then display the total cost.

4. **Guessing Game**  
   Write a program where the computer randomly selects a number between 1–20, and the user must guess it. Give hints if the guess is too high or too low.

5. **File Handling**  
   Write a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.


```python
# 1
def grades():
    """
    Function that accepts 5 marks from user, calculates the average
    and grades the average
    ==========================
    args: 
    ==========================
    return: grade
    """
    m = int(input("Please input marks: "))
    n = int(input("Please input marks: "))
    o = int(input("Please input marks: "))
    p = int(input("Please input marks: "))
    q = int(input("Please input marks: "))

    average = (m + n + o + p + q) / 5

    if average >= 80:
        print("You have an A")
    elif average >= 60:
        print("You have a B")
    elif average >= 40:
        print("You have a C")
    else:
        print("You have an F")
```

```python
grades()
```

```python
# 2

def add(x, y):
    return x + y
def subtract(x, y):
    return x - y
def multiply(x, y):
    return x * y
def divide(x, y):
    if y==0:
        return "Error division by 0!"
    return x / y

def calculator():
    print('=====initialising calculator======')

    while True:
        print('\n Select operation')
        print("1. Add (+)")
        print("2. Subtract (-)")
        print("3. Multiply (*)")
        print("4. Divide (/)")
        print("5. Exit")

        selected= input('Enter Selected (1-5): ').strip()

        if selected == '5':
            print("=========Exiting calculator========")
            break

        if selected in ('1', '2', '3', '4'):
            try:
                num1= float(input("Enter your first number: "))
                num2= float(input("Enter your second number: "))
            except ValueError:
                print("Invalid input! Enter numbers only!")
                continue
            
            if selected == '1':
                print(f"Result: {num1} + {num2} = {add(num1, num2)}")
            elif selected == '2':
                print(f"Result: {num1} - {num2} = {subtract(num1, num2)}")
            elif selected == '3':
                print(f"Result: {num1} * {num2} = {multiply(num1, num2)}")
            elif selected == '4':
                print(f"Result: {num1} / {num2} = {divide(num1, num2)}")
            break
        else:
            print("Invalid Selected! Please select a valid operation (1-5).")
```

```python
if __name__ == "__main__":
    calculator()
```

```python
# 12
"""
a for loop is used for iterating over a sequence like a list, string, or range with a known number of iterations, 
whereas a while loop runs continuously as long as a specific condition remains true, and is useful when the number of iterations
is not known beforehand
"""
#for loop
fruits = ["apple", "mango", "guava"]

print("For Loop Output:")
for fruit in fruits:
    print(f"I like {fruit}")

#while loop
count = 3

print("\nWhile Loop Output:")
while count > 0:
    print(f"Countdown: {count}")
    count -= 1

print("Hurraay!")
```

```python
# 13
"""
exceptions are events that occur during the execution of a program 
that disrupts the programs normal flow of instructions or execution example:"""

def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print("Error: You cannot divide by zero!")
    else:
        print(f"The result is: {result}")
```
