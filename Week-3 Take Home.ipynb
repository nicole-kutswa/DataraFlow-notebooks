```python
# 1
def students():
    """
    ------------------------------------------------------
    Function that stores list of student names to 
    a json file and reads the file back as a python object
    ------------------------------------------------------
    """
    student_details= [{
         "name": input("Please enter your name: "),
         "age": input("Please enter your age: "),
         "grade": input("Please enter your grade: ")
    }]
       
    with open ("student.json", 'w') as file:
        json.dump(student_details, file, indent= 4) 

    with open("student.json", "r") as file:
        student_details= json.load(file)
    return student_details
students()

```

```python
# 2

import random
def math_quiz():
    """
    -----------------------------------------------------------------
    Function that generates random numbers and a random operator
    , asks user for the answer, keeps score, and returns final score.
    -----------------------------------------------------------------
    """
    score= 0
    for i in range(5):
        num1= random.randint(1, 10)
        num2= random.randint(1, 10)

        operation= random.choice(["+","-","*","/",])
        if operation == "+":
            answer= num1 + num2
        elif operation == "-":
            answer= num1 - num2
        elif operation == "*":
            answer= num1 * num2
        elif operation == "/":
            answer= num1 / num2

        question= float(input(f"what is {num1} {operation} {num2}"))
        if question == answer:
            print("Awesome! correct!")
            score += 1

        else:
            print("Wrong!")
        print(f"You scored {score}/5")

# print(f"You scored {score}/5")
math_quiz()

    

```

```python
from datetime import date, datetime

def birthday_countdown():
    """
    -----------------------------------------------
    Counts down days till users next birthday
    -----------------------------------------------
    """

    birthday = input("Enter your birthday (DD/MM): ")

    birthday = datetime.strptime(birthday, "%d/%m")

    today = date.today()

    next_birthday = date(
        today.year,
        birthday.month,
        birthday.day
    )

    if next_birthday < today:
        next_birthday = date(
            today.year + 1,
            birthday.month,
            birthday.day
        )

    days_left = (next_birthday - today).days

    print(f"There are {days_left} days until your next birthday!")


birthday_countdown()
```
