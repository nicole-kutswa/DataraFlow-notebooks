## SQL Assessment (Intermediate SQL)

This part tests what you learnt in **`4-Intermediate SQL/Intermediate SQL.ipynb`**. Every task runs against the
**live cohort database** using your **read-only** account.

**How it works**
1. Run the **SQL setup** cell below first (and again after restarting the kernel).
2. In each task cell, write your query **between the triple quotes**, then run the cell.
3. `answer(...)` runs your query, shows the result, and tells you if it's ✅ correct. If not, it tells you
   whether the **shape** (rows × columns) is right, so you know where to look.
4. Match the **column order**, **sort order** and **rounding** asked for. Column *names* aren't checked,
   but use the names given.
5. Run `sql_score()` at the end to see your SQL score.

Writes are blocked: the database is read-only, so build everything inside your `SELECT` (CTEs are your workspace).

````python

import hashlib
import json
import datetime
import math
import re

import pandas as pd


_DATE_PATTERN = re.compile(r"^\d{4}-\d{1,2}-\d{1,2}")


def _norm(value):
    """Normalize values so SQL results can be fingerprinted consistently."""

    if value is None:
        return None

    if isinstance(value, (datetime.date, datetime.datetime, pd.Timestamp)):
        return pd.Timestamp(value).date().isoformat()

    if isinstance(value, str):
        value = value.strip()

        if _DATE_PATTERN.match(value):
            try:
                return pd.Timestamp(value).date().isoformat()
            except (ValueError, TypeError):
                pass

        return value

    try:
        if math.isnan(value):
            return None
    except (TypeError, ValueError):
        pass

    try:
        return round(float(value), 2)
    except (TypeError, ValueError, OverflowError):
        return str(value)


def _fingerprint(df):
    """Create a consistent fingerprint of the returned rows."""

    rows = [
        [_norm(value) for value in row]
        for row in df.itertuples(index=False, name=None)
    ]

    return hashlib.sha256(
        json.dumps(rows).encode("utf-8")
    ).hexdigest()


class Practice:

    def __init__(self, expected, titles):
        self.expected = expected
        self.titles = titles

        # None = not attempted, True = correct, False = incorrect
        self.results = {
            task_id: None for task_id in expected
        }

    def sql(self, task_id, query_str):
        """Run a query and check its shape and fingerprint."""

        from db.dataraflow_db import run_query

        if task_id not in self.expected:
            raise KeyError(f"Unknown task ID: {task_id}")

        print(
            f"\nRunning {task_id}: "
            f"{self.titles.get(task_id, '')}"
        )

        if not isinstance(query_str, str) or not query_str.strip():
            self.results[task_id] = None
            print("⏳ Not attempted yet. Please provide a SQL query.")
            return

        try:
            df = run_query(query_str)

        except Exception as error:
            self.results[task_id] = False
            print(
                f"❌ Query execution failed: "
                f"{type(error).__name__}: {str(error)[:300]}"
            )
            return

        expected_shape, expected_hash = self.expected[task_id]

        # Check the number of rows and columns.
        if list(df.shape) != expected_shape:
            self.results[task_id] = False

            print(
                f"❌ Incorrect shape. Expected "
                f"{expected_shape[0]} row(s) × "
                f"{expected_shape[1]} column(s), "
                f"but received {df.shape[0]} × {df.shape[1]}."
            )

            try:
                display(df.head(15))
            except NameError:
                print(df.head(15))

            return

        # Check the actual data, not just the shape.
        actual_hash = _fingerprint(df)
        is_correct = actual_hash == expected_hash

        self.results[task_id] = is_correct

        try:
            display(df.head(15))
        except NameError:
            print(df.head(15))

        if is_correct:
            print(f"✅ {task_id}: Correct!")
        else:
            print(
                f"❌ {task_id}: Incorrect answer. "
                "The shape is correct, but the data fingerprint "
                "does not match."
            )
            print(
                "Check the selected columns, sorting, date conversion, "
                "rounding, filters, and NULL handling."
            )

    def score(self):
        """Display results and calculate the total score."""

        rows = []

        for task_id in self.expected:
            result = self.results[task_id]

            if result is True:
                status = "Correct"
            elif result is False:
                status = "Incorrect"
            else:
                status = "Not attempted"

            rows.append([
                task_id,
                self.titles.get(task_id, ""),
                status
            ])

        results_df = pd.DataFrame(
            rows,
            columns=["Task", "Topic", "Result"]
        )

        print("\n========== SQL PRACTICE SCORE ==========")
        print(results_df.to_string(index=False))

        correct = sum(
            result is True for result in self.results.values()
        )
        total = len(self.expected)
        attempted = sum(
            result is not None for result in self.results.values()
        )

        percentage = (correct / total * 100) if total else 0.0

        print(f"\nCorrect answers: {correct}")
        print(f"Tasks attempted: {attempted}/{total}")
        print(f"Total score: {correct}/{total}")
        print(f"Percentage: {percentage:.1f}%")

        return {
            "correct": correct,
            "total": total,
            "attempted": attempted,
            "percentage": round(percentage, 1),
            "results": results_df
        }

    def sql_score(self):
        """Alias for score(), matching your previous assessment."""

        return self.score()
````

### SQL Task 2 · Filtering & NULLs
For **London** (`week06.london_2014`), return **one row** with three counts:
1. `no_event_days`: days with **no event recorded**
2. `rain_days`: days whose event mentions **Rain** anywhere (e.g. `Rain`, `Fog-Rain`, `Rain-Thunderstorm`)
3. `wet_but_no_event_days`: days with **no event recorded** but `Precipitationmm > 0`

⚠️ The events column name starts with a **space**: `[ Events]`. And remember, `= NULL` never matches.
*Revisit: §1.2, §3.2, §3.3.*

````python
SQL_02 = """

SELECT
    COUNT(CASE
        WHEN [ Events] IS NULL THEN 1
    END) AS no_event_days,

    COUNT(CASE
        WHEN [ Events] LIKE '%Rain%' THEN 1
    END) AS rain_days,

    COUNT(CASE
        WHEN [ Events] IS NULL
         AND Precipitationmm > 0 THEN 1
    END) AS wet_but_no_event_days

FROM week06.london_2014;

"""
answer("s02", SQL_02)
````

### SQL Task 3 · CASE & conditional aggregation
From `week16.customer_churn_prediction`, for each `Contract_Type` return: number of `customers`, number
`churned` (`Churn = 1`), and `churn_rate_pct` (percentage, **rounded to 1 decimal place**).

**Return:** `Contract_Type`, `customers`, `churned`, `churn_rate_pct`, sorted by `churn_rate_pct` (highest first).
⚠️ Watch out for the integer-division trap. *Revisit: §4.4, §5.2.*

````python
SQL_03 = """
SELECT 
    Contract_Type, 
    COUNT(*) as customers, 
    COUNT(
        CASE WHEN Churn = 1
        THEN 1
        END) as churned,

    ROUND(100.0*
    COUNT(CASE WHEN Churn = 1 THEN 1 END)/COUNT(*), 1) AS churn_rate_pct
FROM week16.customer_churn_prediction
GROUP BY Contract_type
ORDER BY churn_rate_pct DESC
    
"""
answer("s03", SQL_03)
````

### SQL Task 4 · GROUP BY + HAVING
From `week09.comtrade_milk_uk_monthly_14`, find the partner countries the UK **exported** milk to in

⚠️ Exclude the `'World'` total row. *Revisit: §6.2.*

### SQL Task 5 · JOIN across two sources
Join WHO TB data (`week06.who_pop_tb_all`) with World Bank population (`week08.wb_pop_2013`) on the country
name. Compute **TB deaths per 100,000 people** using the **World Bank** population (`[SP.POP.TOTL]`),

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**
for 2014, **rounded to 1 decimal place**. Use exactly these city labels:
`'Beijing'`, `'Brasilia'`, `'Cape Town'`, `'Delhi'`, `'London'`, `'Moscow'`.

**Return:** `city`, `avg_mean_temp_c`, sorted by `city` (A→Z).
⚠️ `Mean TemperatureC` is stored as a whole number, so watch the integer-division trap. *Revisit: §8.1–8.2.*

````python
SQL_07 ="""
WITH all_cities AS (
    SELECT 'London' AS city, [Mean TemperatureC] AS mean_temp
    FROM week06.london_2014

    UNION ALL
    SELECT 'Beijing', [Mean TemperatureC]
    FROM week06.beijing_pek_2014

    UNION ALL
    SELECT 'Brasilia', [Mean TemperatureC]
    FROM week06.brasilia_bsb_2014

    UNION ALL
    SELECT 'Cape Town', [Mean TemperatureC]
    FROM week06.capetown_cpt_2014

    UNION ALL
    SELECT 'Delhi', [Mean TemperatureC]
    FROM week06.delhi_del_2014

    UNION ALL
    SELECT 'Moscow', [Mean TemperatureC]
    FROM week06.moscow_svo_2014
)
SELECT
    city,
    ROUND(AVG(1.0 * mean_temp), 1) AS avg_mean_temp_c
FROM all_cities
GROUP BY city
ORDER BY city ASC;
"""
answer("s07", SQL_07)
````

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
SQL_08 = """
WITH ranked AS (
    SELECT
        TRY_CONVERT(date, GMT) AS day,
        MONTH(TRY_CONVERT(date, GMT)) AS month_no,
        [Max TemperatureC] AS max_temp,
        ROW_NUMBER() OVER (
            PARTITION BY MONTH(TRY_CONVERT(date, GMT))
            ORDER BY
                [Max TemperatureC] DESC,
                TRY_CONVERT(date, GMT) ASC
        ) AS rn
    FROM week06.london_2014
)
SELECT
    month_no,
    day,
    max_temp
FROM ranked
WHERE rn = 1
ORDER BY month_no ASC;
"""
answer("s08", SQL_08)
````

### SQL Task 9 · Running total
From `week09.comtrade_milk_uk_monthly_14`, compute the UK's **total milk exports per month** (exclude `'World'`)
and a **year-to-date running total**.

**Return:** `Period`, `exports_usd`, `ytd_exports_usd`, sorted by `Period`. *Revisit: §11.4–11.5.*

````python
SQL_09 = """
WITH monthly AS (
    SELECT
        Period,
        SUM([Trade Value (US$)]) AS exports_usd
    FROM week09.comtrade_milk_uk_monthly_14
    WHERE [Trade Flow] = 'Exports'
      AND Partner <> 'World'
    GROUP BY Period
)
SELECT
    Period,
    exports_usd,
    SUM(exports_usd) OVER (
        ORDER BY Period
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS ytd_exports_usd
FROM monthly
ORDER BY Period ASC;
"""
answer("s09", SQL_09)
````
